### Question 1

</font>


What is the lexical diversity of the given text input? (i.e. ratio of unique tokens to the total number of tokens)
<br>*This function should return a float.*


```python
def answer_one(text: str) -> float:
    if not text.strip():
        return 0.0

    tokens = nltk.word_tokenize(text)

    total_tokens = len(tokens)
    if total_tokens == 0:
        return 0.0

    unique_tokens = set(tokens)

    return len(unique_tokens) / total_tokens
```

### Question 2

</font>

What percentage of tokens is 'whale'or 'Whale'?
<br>*This function should return a float.*

```python
from nltk.tokenize import word_tokenize
def answer_two(text: str) -> float:
    if not text.strip():
        return 0.0

    tokens = nltk.word_tokenize(text)
    count = tokens.count('whale') + tokens.count('Whale')
    total_tokens = len(tokens)
        
    return (count / total_tokens) * 100
```

### Question 3

</font>

What are the 20 most frequently occurring (unique) tokens in the text? What is their frequency?
<br>*This function should return a list of 10 tuples where each tuple is of the form `(token, frequency)`. The list should be sorted in descending order of frequency.*

```python
from nltk.probability import FreqDist
def answer_three(text: str) -> list:
    if not text.strip():
        return []

    tokens = nltk.word_tokenize(text)
    dist = FreqDist(tokens)

    return dist.most_common(20)
```

### Question 4

</font>

What tokens have a length of greater than 5 and frequency of more than 150?
<br>*This function should return a sorted list of the tokens that match the above constraints. To sort your list, use `sorted()`*

```python
def answer_four(text: str) -> list:
    if not text.strip(): 
        return []

    tokens = nltk.word_tokenize(text)
    dist = FreqDist(tokens)

    matched_tokens = [token for token, freq in dist.items() if len(token) > 5 and freq > 150]

    return sorted(matched_tokens)
```

<br>

### Question 6

</font>

What unique words have a frequency of more than 2000? What is their frequency?
<br>*This function should return a list of tuples of the form `(frequency, word)` sorted in descending order of frequency.*


```python
def answer_six(text: str) -> list:
    if not text.strip(): 
        return []

    tokens = nltk.word_tokenize(text)
    dist = FreqDist(tokens)
    matched_tokens = [(freq, token) for token, freq in dist.items() if freq > 2000]

    return sorted(matched_tokens, reverse=True)
```

### Question 7

</font>

What is the average number of tokens per sentence?
<br>*This function should return a float.*

```python
from nltk.tokenize import sent_tokenize,word_tokenize
import numpy as np 
```

```python
def answer_seven(text: str) -> float:
    if not text.strip():
        return 0.0

    tokens = nltk.word_tokenize(text)
    sentences = nltk.sent_tokenize(text)

    avg = len(tokens) / len(sentences)
    
    return avg
```

### Question 8

</font>

What are the 5 most frequent parts of speech in this text? What is their frequency?
<br>*This function should return a list of tuples of the form `(part_of_speech, frequency)` sorted in descending order of frequency.*

```python
nltk.download('averaged_perceptron_tagger_eng')

def answer_eight(text:str) -> list:
    if not text.strip():
        return []

    tokens = nltk.word_tokenize(text)
    tagged_tokens = nltk.pos_tag(tokens)
    tag_freq = nltk.FreqDist([tag for (word, tag) in tagged_tokens])

    return tag_freq.most_common(5)
```

### Question 9

Create spelling recommender, that take a list of misspelled words and recommends a correctly spelled word for every word in the list.

For every misspelled word, the recommender should find find the word in `correct_spellings` that has the shortest `edit distance` (you may need  to use `nltk.edit_distance(word_1, word_2, transpositions=True)`), and starts with the same letter as the misspelled word, and return that word as a recommendation.

Recommender should provide recommendations for the three words: `['cormulent', 'incendenece', 'validrate']`.
<br>*This function should return a list of length three:
`['cormulent_reccomendation', 'incendenece_reccomendation', 'validrate_reccomendation']`.*

```python
from nltk.corpus import words
```

```python
def answer_nine(default_words= ['cormulent', 'incendenece', 'validrate']):
    correct_spelling = words.words()

    reccomendation = []

    for entry in default_words:
        valid_words = [word for word in correct_spelling if word.startswith(entry[0])]
        best_match = min(
            valid_words,
            key=lambda word: nltk.edit_distance(entry, word, transpositions=True)
        )

        reccomendation.append(best_match)

    return reccomendation
```
